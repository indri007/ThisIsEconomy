### 3.4.1 Pelabelan Sarkasme Berdasarkan Kata Kunci

Untuk pelabelan sarkasme, kita dapat menggunakan pendekatan berbasis kata kunci sederhana. Meskipun sarkasme seringkali bergantung pada konteks dan nada, keberadaan kata-kata atau frasa tertentu dapat menjadi indikator awal. Kita akan membuat daftar kata kunci yang umum digunakan dalam kalimat sarkasme di Bahasa Indonesia. Jika salah satu kata kunci ini ditemukan dalam `clean_text`, maka tweet akan diberi label 'Sarkasme', jika tidak, 'Non-Sarkasme'.

**Catatan:** Pendekatan ini memiliki keterbatasan karena sarkasme sulit dideteksi hanya dengan kata kunci dan sering membutuhkan pemahaman konteks yang lebih dalam. Ini adalah langkah awal dan bisa disempurnakan dengan model yang lebih kompleks nanti.

````python
sarcasm_keywords = ['mantap sekali', 'hebat ya', 'lucu sekali', 'pintar sekali', 'bagus banget']

def label_sarcasm_keyword(text):
    if not isinstance(text, str): return 'Non-Sarkasme'
    return 'Sarkasme' if any(kw in text.lower() for kw in sarcasm_keywords) else 'Non-Sarkasme'

# Label seluruh dataframe utama sebelum split
df['sarcasm_label_keyword'] = df['clean_text'].apply(label_sarcasm_keyword)

from sklearn.model_selection import train_test_split
df_train, df_temp = train_test_split(df, test_size=0.3, random_state=42)
df_val, df_test = train_test_split(df_temp, test_size=0.5, random_state=42)
print("Labeling & Splitting ulang selesai.")
````

### 3.6 Pelabelan Lanjutan dengan Kecerdasan Buatan (Gemini API)

Setelah membagi data, kita akan melanjutkan pelabelan emosi dan sarkasme menggunakan Google Gemini API. Metode ini diharapkan dapat memberikan hasil yang lebih akurat dibandingkan pelabelan berbasis kata kunci sederhana karena Gemini memiliki pemahaman konteks dan nuansa bahasa yang lebih baik.

````python
!pip install -q openai
import pandas as pd
import time
import random
from openai import OpenAI

# 1. Konfigurasi OpenRouter API
OPENROUTER_API_KEY = "REDACTED_API_KEY"
client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=OPENROUTER_API_KEY,
)

# 2. Persiapan Data (10 sampel dari df_test)
df_sample = df_test.head(10).copy()
column_to_use = "clean_text"

# 3. Fungsi Labeling dengan OpenRouter (Model: google/gemini-2.0-flash-001 atau sejenis)
def label_two_tier_with_retry(text, max_retries=5):
    prompt = f"""
    Anda adalah analis media sosial akademik.
    Analisis cuitan Twitter/X berikut terkait program MBG.
    Teks: \"{text}\"

    Tentukan:
    1. Label Emosi utama: [Marah, Sedih, Senang, Takut, Jijik, Percaya, Antisipasi, Kaget, Netral]
    2. Label Sarkasme: [Sarkasme, Non-Sarkasme]
    3. Keterangan singkat maksimal 10 kata.

    Penting: Perhatikan emoji dan nada sindiran.
    Format jawaban WAJIB: Emosi|Sarkasme|Keterangan
    """

    for i in range(max_retries):
        try:
            response = client.chat.completions.create(
                model="google/gemini-2.0-flash-001",
                messages=[{"role": "user", "content": prompt}]
            )
            hasil = response.choices[0].message.content.strip()
            split_hasil = hasil.split("|")
            if len(split_hasil) >= 3:
                return [s.strip() for s in split_hasil[:3]]
            return ["Error", "Error", "Format salah"]

        except Exception as e:
            if "429" in str(e):
                wait_time = (2 ** i) * 15 + random.uniform(1, 5)
                print(f"   (!) Rate limit (429). Menunggu {wait_time:.1f} detik...")
                time.sleep(wait_time)
            else:
                print(f"   (!) Error API: {e}")
                return ["Error", "Error", str(e)]

    return ["Error", "Error", "Max retries reached"]

# 4. Eksekusi Labeling
hasil_emosi, hasil_sarkasme, hasil_keterangan = [], [], []
print("Memulai labeling pada 10 sampel data (OpenRouter - Gemini 2.0 Flash)...\n")

for i, (idx, row) in enumerate(df_sample.iterrows()):
    res = label_two_tier_with_retry(str(row[column_to_use]))

    hasil_emosi.append(res[0])
    hasil_sarkasme.append(res[1])
    hasil_keterangan.append(res[2])

    print(f"[{i+1}/10] Selesai: {res[0]} | {res[1]}")
    time.sleep(2) # Jeda singkat untuk stabilitas

# 5. Simpan & Tampilkan
df_sample["Label_Emosi"] = hasil_emosi
df_sample["Label_Sarkasme"] = hasil_sarkasme
df_sample["Keterangan"] = hasil_keterangan

display(df_sample[[column_to_use, "Label_Emosi", "Label_Sarkasme", "Keterangan"]])
df_sample.to_csv("hasil_labeling_sampel_openrouter.csv", index=False)
print("\n✅ Berhasil disimpan: hasil_labeling_sampel_openrouter.csv")
````

### 4. Pelatihan Model Klasifikasi (IndoBERT)
Pada tahap ini, kita akan menggunakan model pre-trained `indobenchmark/indobert-base-p2` untuk melakukan klasifikasi otomatis berdasarkan data yang telah dilabeli.

````python
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import BertTokenizer
from sklearn.preprocessing import LabelEncoder
import numpy as np

# 1. Persiapan Label
le = LabelEncoder()
le.fit(['Sarkasme', 'Non-Sarkasme'])
df_train['label_idx'] = le.transform(df_train['sarcasm_label_keyword'])
df_val['label_idx'] = le.transform(df_val['sarcasm_label_keyword'])

# 2. Inisialisasi Tokenizer
PRE_TRAINED_MODEL_NAME = 'indobenchmark/indobert-base-p2'
tokenizer = BertTokenizer.from_pretrained(PRE_TRAINED_MODEL_NAME)

class TweetDataset(Dataset):
    def __init__(self, tweets, targets, tokenizer, max_len):
        self.tweets = tweets
        self.targets = targets
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.tweets)

    def __getitem__(self, item):
        # Menggunakan call method tokenizer secara langsung untuk menghindari AttributeError
        encoding = self.tokenizer(
            str(self.tweets[item]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'targets': torch.tensor(self.targets[item], dtype=torch.long)
        }

train_data_loader = DataLoader(TweetDataset(df_train.clean_text.to_numpy(), df_train.label_idx.to_numpy(), tokenizer, 128), batch_size=16)
val_data_loader = DataLoader(TweetDataset(df_val.clean_text.to_numpy(), df_val.label_idx.to_numpy(), tokenizer, 128), batch_size=16)
print('Data Loader siap.')
````

### 5. Inisialisasi Model & Training Loop

````python
import torch
import matplotlib.pyplot as plt
from transformers import BertForSequenceClassification
from torch.optim import AdamW

if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

# Inisialisasi Model
# le.classes_ berasal dari LabelEncoder di cell sebelumnya
model = BertForSequenceClassification.from_pretrained('indobenchmark/indobert-base-p2', num_labels=len(le.classes_)).to(device)
optimizer = AdamW(model.parameters(), lr=2e-5)

# Contoh history untuk visualisasi
history = {'train_acc': [0.55, 0.75, 0.88], 'val_acc': [0.52, 0.70, 0.82]}

plt.figure(figsize=(8, 5))
plt.plot(history['train_acc'], marker='o', label='Train Accuracy')
plt.plot(history['val_acc'], marker='s', label='Validation Accuracy')
plt.title('IndoBERT Fine-Tuning Progress')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()

print(f'Model IndoBERT siap dilatih di device: {device}')
````

### 6. Proses Fine-Tuning IndoBERT
Kita akan menjalankan loop pelatihan untuk memperbarui bobot model berdasarkan label sarkasme yang telah kita siapkan.

````python
import torch.nn as nn
import numpy as np

def train_epoch(model, data_loader, loss_fn, optimizer, device, n_examples):
    model.train()
    losses = []
    correct_predictions = 0
    for d in data_loader:
        input_ids = d["input_ids"].to(device)
        attention_mask = d["attention_mask"].to(device)
        targets = d["targets"].to(device)

        outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        _, preds = torch.max(outputs.logits, dim=1)
        loss = loss_fn(outputs.logits, targets)

        correct_predictions += torch.sum(preds == targets)
        losses.append(loss.item())

        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        optimizer.zero_grad()

    return correct_predictions.double() / n_examples, np.mean(losses)

def eval_model(model, data_loader, loss_fn, device, n_examples):
    model.eval()
    losses = []
    correct_predictions = 0
    with torch.no_grad():
        for d in data_loader:
            input_ids = d["input_ids"].to(device)
            attention_mask = d["attention_mask"].to(device)
            targets = d["targets"].to(device)

            outputs = model(input_ids=input_ids, attention_mask=attention_mask)
            _, preds = torch.max(outputs.logits, dim=1)
            loss = loss_fn(outputs.logits, targets)

            correct_predictions += torch.sum(preds == targets)
            losses.append(loss.item())

    return correct_predictions.double() / n_examples, np.mean(losses)

# Inisialisasi parameter pelatihan
EPOCHS = 3
loss_fn = nn.CrossEntropyLoss().to(device)
train_acc_history, val_acc_history = [], []

for epoch in range(EPOCHS):
    print(f'Epoch {epoch + 1}/{EPOCHS}')
    train_acc, train_loss = train_epoch(model, train_data_loader, loss_fn, optimizer, device, len(df_train))
    val_acc, val_loss = eval_model(model, val_data_loader, loss_fn, device, len(df_val))

    train_acc_history.append(train_acc.item())
    val_acc_history.append(val_acc.item())

    print(f'Train loss {train_loss:.4f} accuracy {train_acc:.4f}')
    print(f'Val loss {val_loss:.4f} accuracy {val_acc:.4f}')

# Visualisasi Hasil
plt.figure(figsize=(10, 5))
plt.plot(train_acc_history, label='Train Accuracy', marker='o')
plt.plot(val_acc_history, label='Validation Accuracy', marker='s')
plt.title('IndoBERT Fine-Tuning Results')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()
````

### 9. Implementasi Multi-Task IndoBERT (Emosi & Sarkasme)
Sesuai dengan rancangan penelitian, kita akan membuat model kustom yang menggabungkan klasifikasi emosi (9 kelas) dan sarkasme (biner) dalam satu arsitektur.

````python
import torch.nn as nn
from transformers import BertModel

# 1. Persiapan Label Encoder untuk Emosi
emotion_list = ['Marah', 'Sedih', 'Senang', 'Takut', 'Jijik', 'Percaya', 'Antisipasi', 'Kaget', 'Netral']
le_emotion = LabelEncoder()
le_emotion.fit(emotion_list)

class MultiTaskIndoBERT(nn.Module):
    def __init__(self, n_emotion_classes):
        super(MultiTaskIndoBERT, self).__init__()
        self.bert = BertModel.from_pretrained('indobenchmark/indobert-base-p2')
        self.drop = nn.Dropout(p=0.3)

        # Head 1: Klasifikasi Emosi (Softmax output via CrossEntropy)
        self.emotion_out = nn.Linear(self.bert.config.hidden_size, n_emotion_classes)

        # Head 2: Deteksi Sarkasme (Sigmoid output via BCE atau 2-class Softmax)
        # Sesuai permintaan: Sigmoid untuk biner
        self.sarcasm_out = nn.Linear(self.bert.config.hidden_size, 1)

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled_output = outputs.pooler_output
        output = self.drop(pooled_output)

        emotion_logits = self.emotion_out(output)
        sarcasm_logits = self.sarcasm_out(output)

        return emotion_logits, sarcasm_logits.squeeze(1)

# Inisialisasi Model
if torch.backends.mps.is_available():
    device = torch.device("mps")
elif torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")
mt_model = MultiTaskIndoBERT(n_emotion_classes=len(emotion_list)).to(device)
print("Arsitektur Multi-Task IndoBERT siap.")
````

### 10. Persiapan Data Multi-Task
Kita perlu memperbarui `Dataset` agar mengembalikan label emosi dan sarkasme sekaligus.

````python
class MultiTaskDataset(Dataset):
    def __init__(self, tweets, emotions, sarcasms, tokenizer, max_len):
        self.tweets = tweets
        self.emotions = emotions
        self.sarcasms = sarcasms
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.tweets)

    def __getitem__(self, item):
        encoding = self.tokenizer(
            str(self.tweets[item]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_tensors='pt'
        )
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'emotion_label': torch.tensor(self.emotions[item], dtype=torch.long),
            'sarcasm_label': torch.tensor(self.sarcasms[item], dtype=torch.float)
        }

# Fungsi pembersihan untuk menangani label yang tidak terduga dari API
def clean_ai_labels(label):
    if label in emotion_list:
        return label
    return 'Netral' # Default jika output API tidak sesuai kategori

# Terapkan pembersihan pada data yang sudah di-label AI
df_train_ai['ai_emotion'] = df_train_ai['ai_emotion'].apply(clean_ai_labels)
df_val_ai['ai_emotion'] = df_val_ai['ai_emotion'].apply(clean_ai_labels)

# Memetakan label string ke numerik
df_train_ai['emotion_idx'] = le_emotion.transform(df_train_ai['ai_emotion'])
df_train_ai['sarcasm_idx'] = df_train_ai['ai_sarcasm'].apply(lambda x: 1.0 if x == 'Sarkasme' else 0.0)

# Buat DataLoader untuk Multi-Task Learning
mt_train_loader = DataLoader(
    MultiTaskDataset(
        df_train_ai.clean_text.to_numpy(),
        df_train_ai.emotion_idx.to_numpy(),
        df_train_ai.sarcasm_idx.to_numpy(),
        tokenizer,
        128
    ),
    batch_size=16
)

print("Data Loader Multi-Task siap dengan label yang telah divalidasi.")
````

### 11. Pelatihan Multi-Task IndoBERT (Fine-Tuning)
Pada tahap ini, model akan belajar mengenali 9 jenis emosi dan mendeteksi sarkasme secara simultan.

````python
import torch.optim as optim
import torch.nn.functional as F

# Definisi Loss Functions
loss_fn_emotion = nn.CrossEntropyLoss().to(device)
loss_fn_sarcasm = nn.BCEWithLogitsLoss().to(device)

# Optimizer sesuai metodologi (AdamW, lr=2e-5)
optimizer = optim.AdamW(mt_model.parameters(), lr=2e-5)

def train_multi_task(model, data_loader, optimizer, device, n_examples):
    model.train()
    losses = []
    correct_emotion = 0
    correct_sarcasm = 0

    for d in data_loader:
        input_ids = d["input_ids"].to(device)
        attention_mask = d["attention_mask"].to(device)
        emotion_labels = d["emotion_label"].to(device)
        sarcasm_labels = d["sarcasm_label"].to(device)

        emotion_logits, sarcasm_logits = model(input_ids, attention_mask)

        # Hitung Loss
        loss_emo = loss_fn_emotion(emotion_logits, emotion_labels)
        loss_sarc = loss_fn_sarcasm(sarcasm_logits, sarcasm_labels)

        # Total Loss (Bisa ditambahkan bobot jika perlu)
        total_loss = loss_emo + loss_sarc

        losses.append(total_loss.item())

        # Akurasi Emosi
        _, emo_preds = torch.max(emotion_logits, dim=1)
        correct_emotion += torch.sum(emo_preds == emotion_labels)

        # Akurasi Sarkasme
        sarc_preds = torch.round(torch.sigmoid(sarcasm_logits))
        correct_sarcasm += torch.sum(sarc_preds == sarcasm_labels)

        total_loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        optimizer.zero_grad()

    return correct_emotion.double() / n_examples, correct_sarcasm.double() / n_examples, np.mean(losses)

# Jalankan Pelatihan (Contoh: 5 Epoch sesuai rancangan)
EPOCHS = 5
print("Memulai Multi-Task Training...")
for epoch in range(EPOCHS):
    emo_acc, sarc_acc, loss = train_multi_task(mt_model, mt_train_loader, optimizer, device, len(df_train_ai))
    print(f"Epoch {epoch+1}/{EPOCHS} | Loss: {loss:.4f} | Emo Acc: {emo_acc:.4f} | Sarc Acc: {sarc_acc:.4f}")

print("\nPelatihan Multi-Task selesai.")
````

### 15. Social Network Analysis (SNA)
In this section, we build a network graph where **Nodes** represent users and **Edges** represent interactions (Mentions). We will then calculate the following metrics:
1.  **Degree Centrality**: To measure user popularity.
2.  **Betweenness Centrality**: To identify information brokers.
3.  **Eigenvector Centrality**: To identify influential users connected to other influential users.
4.  **Modularity (Louvain Community Detection)**: To detect clusters/communities.

````python
!pip install python-louvain
import networkx as nx
import community.community_louvain as community_louvain
import matplotlib.pyplot as plt
import re
import pandas as pd

# 1. Menyiapkan Data Jaringan (Edges)
def extract_mentions(text):
    return re.findall(r'@(\w+)', text)

# Membangun daftar interaksi (source: username, target: mention)
edges = []
for _, row in df.iterrows():
    source = row['username']
    mentions = extract_mentions(row['full_text'])
    for target in mentions:
        edges.append((source, target))

# 2. Inisialisasi Graph
G = nx.Graph()
G.add_edges_from(edges)

print(f"Jaringan berhasil dibangun dengan {G.number_of_nodes()} simpul dan {G.number_of_edges()} hubungan.")

# 3. Kalkulasi Metrik SNA
degree_cent = nx.degree_centrality(G)
betweenness_cent = nx.betweenness_centrality(G)
eigen_cent = nx.eigenvector_centrality(G, max_iter=1000)

# 4. Deteksi Komunitas (Modularity)
# Menggunakan community_louvain yang diimpor secara spesifik
partition = community_louvain.best_partition(G)

# Simpan hasil ke DataFrame
sna_data = []
for node in G.nodes():
    sna_data.append({
        'username': node,
        'degree_centrality': degree_cent[node],
        'betweenness_centrality': betweenness_cent[node],
        'eigenvector_centrality': eigen_cent[node],
        'community': partition[node]
    })

df_sna = pd.DataFrame(sna_data).sort_values(by='degree_centrality', ascending=False)

print("\n=== Top 10 Aktor Berpengaruh (Degree Centrality) ===")
display(df_sna.head(10))

# 5. Visualisasi Jaringan (Sederhana)
plt.figure(figsize=(12, 12))
pos = nx.spring_layout(G, k=0.15, iterations=20)
cmap = plt.get_cmap('viridis')

# Gambar simpul berdasarkan komunitas
nodes = nx.draw_networkx_nodes(G, pos, node_size=50, cmap=cmap,
                               node_color=list(partition.values()))
nx.draw_networkx_edges(G, pos, alpha=0.2)

plt.title("Visualisasi Klaster Komunitas pada Jaringan Interaksi MBG")
plt.axis('off')
plt.show()
````

````python
# LANGKAH TAMBAHAN: Fine-tuning IndoBERT untuk 9 Emosi
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments

# 1. Load Dataset berlabel (menggunakan indobert_9_emosi.csv)
df = pd.read_csv('../data/results/indobert_9_emosi.csv')
df = df.dropna(subset=['processed_text', 'predicted_emotion'])

# 2. Encode Labels (Membuat mapping label ke ID)
labels = df['predicted_emotion'].unique().tolist()
label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}
df['label_id'] = df['predicted_emotion'].map(label2id)
print(f"Ditemukan {len(labels)} label unik: {labels}")

# 3. Train-Test Split (80% Train, 20% Validation)
train_texts, val_texts, train_labels, val_labels = train_test_split(
    df['processed_text'].tolist(), 
    df['label_id'].tolist(), 
    test_size=0.2, 
    random_state=42
)

# 4. Tokenizer IndoBERT
model_name = "indolem/indobert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)

train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=128)
val_encodings = tokenizer(val_texts, truncation=True, padding=True, max_length=128)

# 5. Custom Dataset Class
class EmotionDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)

train_dataset = EmotionDataset(train_encodings, train_labels)
val_dataset = EmotionDataset(val_encodings, val_labels)

# 6. Inisialisasi Model
model = AutoModelForSequenceClassification.from_pretrained(
    model_name, 
    num_labels=len(labels),
    id2label=id2label,
    label2id=label2id
)
model.to(device) # device menggunakan variabel dari Langkah 10 (mps/cuda/cpu)

# 7. Training Arguments
training_args = TrainingArguments(
    output_dir='../results/indobert_finetuned',
    num_train_epochs=3,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    warmup_steps=500,
    weight_decay=0.01,
    logging_steps=50,
    evaluation_strategy="epoch",
    save_strategy="epoch",
    # Menggunakan MPS jika tersedia di Mac
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset
)

# 8. Eksekusi Fine-tuning
print("Memulai proses training...")
trainer.train()
````
